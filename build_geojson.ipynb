```python
import pandas as pd
import json
```

```python
with open("data/sections.json") as f:
    data = json.loads(f.read())
```

The data is keyed by РИК. Let's explore the first RIK

```python
first_rik = data['data']['1']

len(first_rik), first_rik[0]
```

```python
def convert_entry(rik, entry):
    rik = int(rik)
    return {
        "type":"Feature",
        "geometry": {
            "type": "Point",
            "coordinates" : [entry['lng'], entry['lat']]
        },
        "properties": {
            "rik": rik,
            "abroad": rik == 32,
            "address": entry['a'].replace('<br />', ', '),
            "sections": ",".join(entry['s']),
            'n_sections': len(entry['s'])
        },
    }
convert_entry(1, first_rik[0])
```

```python
res = []
for rik, entries in data['data'].items():
    for e in entries:
        res.append(convert_entry(rik, e))
```

```python
final = {
    "type": "FeatureCollection",
    "features": res
}
```

```python
with open('web/locations.geojson', 'w') as o:
    o.write(json.dumps(final))
```
